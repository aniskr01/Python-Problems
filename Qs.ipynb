1. Write a single-line list comprehension that takes a list of strings, filters out strings shorter than 4 characters, and converts the remaining strings to uppercase.

```python
words = ["apple","bag","cherry","dog","elderbelly"]
print(f"Original List: {words}")
WORDS = [word.upper() for word in words if len(word)>=4]
print(f"Updated List: {WORDS}")
```

2. Write a function that merges two dictionaries. If a key exists in both dictionaries, sum their values. If a key exists in only one, include it as is.

```python
def merge_dict(dict_a: dict, dict_b: dict) -> dict:
    new_dict = dict_a.copy()

    for key,value in dict_b.items():
        new_dict.update({key: (new_dict.get(key, 0) + value)})

    return new_dict

dict_a = {'a': 10, 'b': 20, 'c': 30}
dict_b = {'b': 10, 'c': 20, 'd': 30}

print(f"Original Dictionary 1: {dict_a}")
print(f"Original Dictionary 2: {dict_b}")
print(f"Merged Dictionary: {merge_dict(dict_a, dict_b)}")
```
